```python
df = df.drop_duplicates()
```

```python
print(df['Product'].unique())
print(df['Category'].unique())
print(df['City'].unique())
```

```python
df['Product'] = df['Product'].str.strip()
df['Category'] = df['Category'].str.strip()
df['City'] = df['City'].str.strip()
```

```python
df = pd.read_excel("sales_data_2000_rows.xlsx")
```

```python
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())
```

```python
print(df.isnull().sum())
```

```python
print("Invalid prices:", (df['Price'] <= 0).sum())
```

```python
print("Invalid prices:", (df['Price'] <= 0).sum())
print(df.shape)
```

```python
df["Revenue"] = df["Quantity"] * df["Price"]
```

```python
df[["Product", "Quantity", "Price", "Revenue"]].head()
```

```python
total_revenue = df["Revenue"].sum()

print("Total Revenue:", total_revenue)
```

```python
product_sales = df.groupby("Product")["Quantity"].sum()

print(product_sales)
```

```python
print("Best Selling Product:", product_sales.index[0])
```

```python
city_revenue = df.groupby("City")["Revenue"].sum().sort_values(ascending=False)

print(city_revenue)
```

```python
print("Best Performing City:", city_revenue.index[0])
```

```python
category_revenue = df.groupby("Category")["Revenue"].sum().sort_values(ascending=False)

print(category_revenue)
```

```python
df["Month"] = df["Date"].dt.to_period("M")
```

```python
monthly_revenue = df.groupby("Month")["Revenue"].sum()

print(monthly_revenue)
```

```python
import matplotlib.pyplot as plt 
```

```python
plt.figure(figsize=(10, 6))

plt.bar(city_revenue.index, city_revenue.values)

plt.title("Revenue by City")
plt.xlabel("City")
plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 8))

plt.pie(
    category_revenue.values,
    labels=category_revenue.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Revenue Contribution by Category")

plt.show()
```

```python
monthly_revenue = df.groupby("Month")["Revenue"].sum().sort_index()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    monthly_revenue.index.astype(str),
    monthly_revenue.values,
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
